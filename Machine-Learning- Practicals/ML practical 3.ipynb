```python
 # Import library 
import matplotlib.pyplot as plt
```

```python
 # Define the function and its derivative 
def df(x):   
    return (x + 3)**2 
    def df(x):    
        return 2 * (x + 3)
```

```python
# Initialize parameters
x_old = 2          # Starting point 
learning_rate = 0.1 
precision = 0.00001 
max_iterations = 1000
```

```python
def f(x):
    return x**2

def gradient(x):
    return 2*x

learning_rate = 0.1
precision = 0.00001
x_old = 10

x_values = [x_old]
y_values = [f(x_old)]

while True:
    x_new = x_old - learning_rate * gradient(x_old)

    x_values.append(x_new)
    y_values.append(f(x_new))

    if abs(x_new - x_old) < precision:
        break

    x_old = x_new

print("Minimum value of x:", x_new)
print("Minimum value of f(x):", f(x_new))
```

```python
# Cell 2: gradient descent
x_values = []
y_values = []

for i in range(max_iterations):
    gradient = df(x_old)
    x_new = x_old - learning_rate * gradient
    x_values.append(x_new)
    y_values.append(f(x_new))
    if abs(x_new - x_old) < precision:
        break
    x_old = x_new

print(f"Local minima occurs at x = {x_new:.5f}")
print(f"Minimum value of function y = {f(x_new):.5f}")
```

```python
# Plotting
plt.plot(x_values, y_values, 'b-o', label='Gradient Descent Path')

plt.title("Gradient Descent Convergence")
plt.xlabel("X")
plt.ylabel("f(x) = (x + 3)^2")

plt.legend()
plt.grid(True)
plt.show()
```
